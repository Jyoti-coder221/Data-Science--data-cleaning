## Cleaning a messy dataset in Pandas, handling missing values, filtering rows, creating new columns

```python
import pandas as pd
df = pd.read_csv("data (1).csv")
df
```

### Handling missing values

```python
print(df.isnull().sum())
```

#### fill missing values with average age

```python
df["Date"] = df["Date"].fillna("Unknown")
```

```python
df["Calories"] = df["Calories"].fillna(df["Calories"].mean())
```

#### filter rows where pulse is greater than 100

```python
high_pulse = df[df["Pulse"] > 100]

print(high_pulse)
```

```python
print(df.isnull().sum())
```

### Create a new column 

#### Create a new column based on Pulse values.
 #####  Pulse greater than 100 is categorized as "High",
 #####  otherwise it is categorized as "Normal"

.

```python
df["Pulse_Category"] = df["Pulse"].apply(
    lambda x: "High" if x > 100 else "Normal"
)

print(df)
```

### Clean dataset

```python
print(df.isnull().sum())
print("|Dataset with New Column:")
print(df)
```

### Conclusion: 
#### The dataset was successfully cleaned by handling missing values, filtering rows, and creating a new column for Pulse classification.
